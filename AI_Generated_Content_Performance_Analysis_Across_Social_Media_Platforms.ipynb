## Stage 2 – Data Cleaning and Pre-processing

Handle missing values (impute)

1. Missing values were identified in Content Type, Comments, and Shares columns.
2. Missing values in Content Type were handled using Mode Imputation.
3. Missing values in Comments and Shares were handled using Median Imputation.
4. Duplicate records were checked and no duplicates were found.
5. The dataset was successfully cleaned and prepared for exploratory data analysis.

```python
#  Missing Value Imputation
```

```python
# Categorical Column Handle(Content Type)
#mode
```

```python
df["Content_Type"]=df["Content_Type"].fillna(df["Content_Type"].mode()[0])
```

```python
df['Content_Type'].isnull().sum()
```

```python
#Numerical Columns Handle(comments,shares)
#median
```

```python
df['Comments']=df['Comments'].fillna(df['Comments'].median())
```

```python
df['Shares']=df['Shares'].fillna(df['Shares'].median())
```

```python
df[["Comments","Shares"]].isnull().sum()
```

```python
df.isnull().sum()
```

```python
#Pie Chart(Ai Type vs Revenue)
```

```python
df.groupby("AI_Type")["Revenue_USD"].mean().sort_values(ascending=False)
```

```python
ai_revenue = df.groupby("AI_Type")["Revenue_USD"].mean()

colors = ["#FF9999", "#66B3FF", "#99FF99", "#FFD700"]

explode = (0,0,0.08,0)

plt.figure(figsize=(7,7))

plt.pie(
    ai_revenue,
    labels=ai_revenue.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors,
    explode=explode,
    shadow=True
)

plt.title("Average Revenue by AI Type", color="darkblue", fontsize=14)

plt.show()
```

#Interpretation
* Explain the chart:

 This pie chart shows the average revenue contribution of each AI content type.

```python
#Correlation Heatmap
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df[['Production_Cost_USD','Views','Likes','Comments','Shares',
        'Engagement_Rate','Watch_Time_Avg_Sec',
        'Retention_Rate','Revenue_USD']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap of Numeric Features", color='darkblue', fontsize=14)
plt.xticks(color="purple")
plt.yticks(color="purple")
plt.show()
```

## Interpretation

* Explain the chart

This correlation heatmap shows the relationship between the numerical features in the dataset. Correlation values range from -1 to +1, where values closer to +1 indicate a strong positive relationship and values closer to -1 indicate a negative relationship.
* What is it saying?

Views and Likes have a very strong positive correlation (0.93).
Views and Comments also have a strong positive correlation (0.92).
Likes and Shares are strongly correlated (0.85).
Production Cost and Watch Time show a strong positive correlation (0.88).
Revenue has a moderate positive relationship with Views (0.57).
* What features you used?

Production_Cost_USD, Views, Likes, Comments, Shares, Engagement_Rate, Watch_Time_Avg_Sec, Retention_Rate, Revenue_USD
* From this what are you showing us?

The heatmap shows that content receiving more views also tends to receive more likes, comments, and shares. It also indicates that watch time is strongly related to production cost, while revenue is moderately influenced by views.

# Business Insight
The correlation heatmap helps identify which performance metrics are strongly related. Features with a high positive correlation, such as Views, Likes, Comments, and Revenue, indicate that improving one metric can positively influence the others. Businesses can use these insights to focus on strategies that increase engagement and maximize revenue
